```python
#importing required libraries
import sqlite3
import pandas as pd
from IPython.display import display
```

```python
# Create a connection to the SQLite database (1 points)
conn = sqlite3.connect('NetFlix.db')
c = conn.cursor()
```

```python
# Create a connection to the SQLite database (1 points)
conn2 = sqlite3.connect('spotify_top50_2021.db')
c = conn.cursor()
```

```python
# Load data from CSV files (1 points)
spotify_df = pd.read_csv("spotify_top50_2021.csv")
netflix_df = pd.read_csv('NetFlix.csv')
```

```python
# Create tables and insert data
spotify_df.to_sql('spotify_songs', conn, if_exists='replace', index=False)
netflix_df.to_sql('netflix_shows', conn, if_exists='replace', index=False)
```


## Q1. Given a Spotify dataset containing various metrics for songs, retrieve the top 5 songs ranked by their popularity. (1 points)

```python
c.execute('''select track_name, artist_name, popularity
from spotify_songs
order by popularity desc
limit 5;

''')

df = pd.DataFrame(c.fetchall(), columns=['Track', 'Artist', 'Popularity'])
display(df)
```


##Q2. Identify and list all songs from the dataset that have a danceability score exceeding 0.7. Ensure that each song included meets this criterion based on its danceability rating. (1 points)

```python
c.execute('''select track_name, artist_name, danceability
from spotify_songs
where danceability > 0.7;

''')
df = pd.DataFrame(c.fetchall(), columns=['Track', 'Artist', 'Danceability'])
display(df)
```

##Q3. Given a dataset from Spotify that includes various artists and their popularity scores, determine the average popularity score for each artist. This will help understand how popular each artist is on average across the dataset. (1 points)

```python
c.execute('''select artist_name, avg(popularity)
from spotify_songs
group by artist_name;

''')
df = pd.DataFrame(c.fetchall(), columns=['Artist', 'Avg Popularity'])
display(df)
```

##Q4. Identify artists who have more than one song in the top 50 charts and count how many songs each of these artists have in the top 50. (2 points)

```python
c.execute('''select artist_name, count(*) as song_count
from spotify_songs
group by artist_name having song_count > 1
;

''')
df = pd.DataFrame(c.fetchall(), columns=['Artist', 'Song_Count'])
display(df)
```

##Q5. Identify and list all the unique genres present in the Netflix dataset to understand the variety of content offered. (1 points)

```python
c.execute(''' select distinct genres
from netflix_shows

''')
df = pd.DataFrame(c.fetchall(), columns=['Genre'])
display(df)
```

##Q6. Find all Netflix shows with "Blue" in the title. (1 points)

```python
c.execute(''' select title, genres
from netflix_shows
where title LIKE '%Blue%';

''')
df = pd.DataFrame(c.fetchall(), columns=['Title', 'Genre'])
display(df)
```


##Q7. Find Spotify songs by Olivia Rodrigo or Dua Lipa. (2 points)

##Q8. Find songs on Spotify dataset that have a tempo between 100 and 120 beats per minute. (1 points)

```python
c.execute('''
select track_name, artist_name, tempo
from spotify_songs
where tempo > 100 AND tempo < 120;
''')
df = pd.DataFrame(c.fetchall(), columns=['Track', 'Artist', 'Tempo'])
display(df)
```

##Q9. Sort a list of Spotify songs first by their popularity in descending order, and then by their danceability in ascending order. (1 points)

```python
c.execute('''select track_name, artist_name, popularity, danceability
from spotify_songs
order by popularity desc, danceability asc;

''')
df = pd.DataFrame(c.fetchall(), columns=['Track', 'Artist', 'Popularity', 'Danceability'])
display(df)
```

## Q10. Identify Netflix shows that have a runtime longer than the average runtime of all shows on the platform. (2 points)

```python
c.execute(''' select title, duration
from netflix_shows
where duration > (select avg(duration)
from netflix_shows);

''')
df = pd.DataFrame(c.fetchall(), columns=['Title', 'Runtime'])
display(df)
```

## Q11. Combine the titles of all Spotify songs and Netflix shows into a single list. Use the UNION operation to merge these two datasets, ensuring that each title appears only once in the final list. (2 points)

```python
c.execute(''' select title from netflix_shows
UNION ALL
select track_name from spotify_songs;

''')
df = pd.DataFrame(c.fetchall(), columns=['Title'])
display(df)
```

## Q12. Create a list where each Spotify song is paired with each Netflix show genre. The goal is to generate combinations of songs and genres to explore how different music might fit with various types of TV shows. (1 points)

```python
c.execute(''' select track_name, artist_name, genres
from spotify_songs, netflix_shows;

''')
df = pd.DataFrame(c.fetchall(), columns=['Track', 'Artist', 'Genre'])
display(df)
```

##Q13. Determine the average popularity of Spotify songs for each genre of Netflix shows by joining the Spotify songs and Netflix shows datasets. Use an INNER JOIN to match records where genres align. (2 points)

```python
c.execute('''select genres, avg(popularity)
from netflix_shows
INNER JOIN spotify_songs
on netflix_shows.title = spotify_songs.track_name
group by genres

''')
df = pd.DataFrame(c.fetchall(), columns=['Genre', 'Popularity'])
print(df)
```
