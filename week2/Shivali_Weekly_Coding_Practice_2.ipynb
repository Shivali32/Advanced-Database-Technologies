```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect('emp.db')
c = conn.cursor()

c.execute('DROP TABLE IF EXISTS EMP')
c.execute('''CREATE TABLE EMP
(
   EMPNO INT PRIMARY KEY NOT NULL,
   ENAME CHAR(50) NOT NULL,
   JOB   CHAR(50) NOT NULL,
   MGR   INT,
   HIREDATE DATE,
   SAL   DECIMAL(10, 2),
   COMM  DECIMAL(10, 2),
   DEPTNO INT
);
''')
```

```python
emp_df = pd.read_csv('emp.csv')
emp_df.to_sql('EMP', conn, if_exists='append', index=False)

c.execute('SELECT * FROM emp')
print(c.fetchall())
```

```python
#  Returning Query Results in a Specified Order
c.execute(''' select ENAME, JOB, SAL
from emp
where DEPTNO = 10
order by sal desc;

''')

df = pd.DataFrame(c.fetchall(), columns=['ENAME', 'JOB', 'SAL'])
display(df)
```

```python
# Returning Query Results in a Specified Order
c.execute(''' select ENAME, JOB, SAL
from emp
where DEPTNO = 10
order by sal asc;

''')

df = pd.DataFrame(c.fetchall(), columns=['ENAME', 'JOB', 'SAL'])
display(df)
```

```python
#  Returning Query Results in a Specified Order
c.execute(''' select ENAME, JOB, SAL
from emp
where DEPTNO = 10
order by 3 desc;

''')

df = pd.DataFrame(c.fetchall(), columns=['ENAME', 'JOB', 'SAL'])
display(df)
```

```python
# Sorting by Multiple Fields
c.execute(''' select EMPNO, DEPTNO, SAL, ENAME, JOB
from emp
order by DEPTNO, SAL desc;

''')

df = pd.DataFrame(c.fetchall(), columns=['ENAME', 'DEPTNO', 'SAL', 'ENAME', 'JOB'])
display(df)
```

```python
# Sorting by Substrings
c.execute(''' select ENAME, JOB
from emp
order by substr(JOB, length(JOB)-1)

''')

df = pd.DataFrame(c.fetchall(), columns=['ENAME', 'JOB'])
display(df)
```

```python
# View
c.execute(''' create view V1
as
select ENAME||' '||DEPTNO as data
from emp;

''')

c.execute('''select * from V2''')

df = pd.DataFrame(c.fetchall(), columns=['DATA'])
display(df)
```

```python
# Null String handling
c.execute('''select ENAME, SAL, COMM
from emp
order by 3;
''')

df = pd.DataFrame(c.fetchall(), columns=['ENAME','SAL','COMM'])
display(df)
```

```python
# case expression for Null handling
c.execute('''select ENAME, SAL, COMM
from (select ENAME,SAL,COMM,case when COMM is null then 0 else 1 end as is_null
from emp) x
order by is_null desc,COMM
''')

df = pd.DataFrame(c.fetchall(), columns=['ENAME','SAL','COMM'])
display(df)
```

```python
# Null at end
c.execute('''select ENAME, SAL, COMM
from (select ENAME,SAL,COMM,case when COMM is null then 0 else 1 end as is_null
from emp) x
order by is_null desc,COMM desc
''')

df = pd.DataFrame(c.fetchall(), columns=['ENAME','SAL','COMM'])
display(df)
```

```python
# Null at start with sal desc
c.execute('''select ename,sal,comm
from emp
order by comm desc nulls first''')

df = pd.DataFrame(c.fetchall(), columns=['ENAME','SAL','COMM'])
display(df)
```

```python
# Sorting
c.execute('''select ENAME, SAL, JOB, COMM,
case when JOB = 'SALESMAN' then COMM else SAL end as ORDERED
from emp
order by 5;

''')

df = pd.DataFrame(c.fetchall(), columns=['ENAME','SAL', 'JOB','COMM', 'ORDERED'])
display(df)
```
